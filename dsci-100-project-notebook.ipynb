#### Task 5.2 Perform the Analysis (20 points)

- Perform the proposed analysis.
- Be sure to comment your code throughout so that it is easy to follow. 

*You may add more code cells if needed.* 


</div>

```r
library(tidymodels) 
# load necessary library

bid_split <- bid_data |>
                initial_split(prop = 0.70, strata = bid) 
# make the initial split

bid_training <- training(bid_split)
bid_testing  <- testing(bid_split) 
# make the testing and training datasets
```

```r
bid_recipe <- recipe(bid ~ conf_rank + conf_wins_pct + reg_wins_pct, data = bid_training) |>
                step_scale(all_predictors()) |>
                step_center(all_predictors())
# make a recipe to standarized and center all predictors

knn_spec <- nearest_neighbor(weight_func = "rectangular", neighbors = tune()) |>
    set_engine("kknn") |>
    set_mode("classification")
# create a model for tuning k

bid_vfold <- vfold_cv(bid_training, v = 5, strata = bid)
# 5 cross-validation splits on the training set

k_values <- tibble(neighbors = seq(from = 1, to = 20, by = 1))
# grid of k values

knn_results <- workflow() |>
    add_recipe(bid_recipe) |>
    add_model(knn_spec) |>
    tune_grid(resamples = bid_vfold, grid = k_values) |>
    collect_metrics()
# create a workflow, cross-validate for each k-value and extract metrics 
```

```r
accuracies_training <- knn_results |>
    filter(.metric == "accuracy")
accuracies_training
# filter for accuracy metrics 

best_k <- accuracies_training |>
    arrange(desc(mean)) |>
    slice(1) |>
    pull(neighbors)
best_k
# identify optimal k value
```

```r
final_knn_spec <- nearest_neighbor(weight_func = "rectangular", neighbors = best_k) |>
    set_engine("kknn") |>
    set_mode("classification")
# create a model using the optimal k-value

final_fit <- workflow() |>
    add_recipe(bid_recipe) |>
    add_model(final_knn_spec) |>
    fit(data = bid_training)
# fit the training set

test_predictions <- predict(final_fit, bid_testing) |>
    bind_cols(bid_testing)
# predict on the testing set

accuracies_testing <- test_predictions |>
    metrics(truth = bid, estimate = .pred_class) |>
    filter(.metric == "accuracy")
accuracies_testing
# calculate the accuracy of the prediction model
```

```r
conf_matrix <- test_predictions |>
    conf_mat(truth = bid, estimate = .pred_class)
conf_matrix
# create the confusion matrix
```
